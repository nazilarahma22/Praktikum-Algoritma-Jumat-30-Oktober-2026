```python

# Fungsi untuk mengubah nilai huruf menjadi angka
def konversi_nilai(huruf, nilai_default=0):
    if huruf == "A":
        return 4
    elif huruf == "B":
        return 3
    elif huruf == "C":
        return 2
    elif huruf == "D":
        return 1
    elif huruf == "E":
        return 0
    else:
        return nilai_default


# Fungsi untuk menghitung rata-rata
def hitung_rata_rata(total, jumlah):
    return total / jumlah


# Program utama
total = 0
jumlah = 0

while True:
    huruf = input("masukkan nilai: ").upper()

    if huruf == "":
        break

    nilai = konversi_nilai(huruf)

    print("nilai =", nilai)

    total += nilai
    jumlah += 1

if jumlah > 0:
    rata = hitung_rata_rata(total, jumlah)
    print("Rata-ratanya adalah:", rata)
```

```python

# Fungsi untuk menentukan akhiran ordinal
def akhiran_ordinal(angka):
    if angka % 100 >= 11 and angka % 100 <= 13:
        return "th"
    elif angka % 10 == 1:
        return "st"
    elif angka % 10 == 2:
        return "nd"
    elif angka % 10 == 3:
        return "rd"
    else:
        return "th"


# Fungsi untuk mengubah angka menjadi ordinal
def ordinal_number(angka):
    akhiran = akhiran_ordinal(angka)
    return (angka, akhiran)


# Program utama
print("Ordinal Number")
print("ketik 0 untuk menghentikan program")

while True:
    angka = int(input("masukkan angka: "))

    hasil = ordinal_number(angka)
    print(hasil)

    if angka == 0:
        print("terima kasih telah menggunakan program saya")
        break
```
